## Core 5 – Scoring how "English-like" a message is (bigram model)

**Model.** From the cleaned Moby Dick text we count

* $\text{frequency}(i)$: how often character $i$ appears,
* $\text{frequency}(ij)$: how often $i$ is immediately followed by $j$ ($27 \times 27 = 729$ pairs),

and set
$$p(i,j) = \frac{\text{frequency}(ij)+1}{\text{frequency}(i)},$$
an estimate of the probability that $i$ is followed by $j$. The $+1$ stops pairs that never occur in Moby Dick from getting probability $0$.

**Score.** For a candidate message $m = i_1 i_2 \dots i_n$,
$$S(m) = \sum_{j=1}^{n-1} \log p(i_j, i_{j+1}).$$

This score is motivated by a first-order Markov model, in which the next character depends only on the current character. Taking natural logarithms turns the product of the weights $p(i_j, i_{j+1})$ into a sum. This avoids numerical underflow (for message 2 the product would be about $10^{-1968}$) and preserves the ranking, because the logarithm is strictly increasing. For candidate decodings of the same ciphertext, a higher score indicates adjacent-character patterns that are more consistent with the training corpus.

$S$ is a large negative number: **the larger (closer to 0) it is, the more English-like the message.**

*Uses from earlier sections:* `moby` (Core 2), `ciphertexts` (Core 1), `plain_alphabet` (Core 4).

We use the cleaned training text from Core 2 to construct the scoring table. Candidate decodings from earlier sections are used to check the scoring function, which will then be used by the Metropolis algorithm in Core 6.

### Step 1 – Training text from Core 2: re-cleaning and checks
We reuse the Moby Dick text downloaded in Core 2. For bigram counts the text must contain **only** the 27 characters
of the cipher alphabet, otherwise some consecutive pairs would not be valid pairs $(i, j)$. So we re-clean the raw text
with the booklet's simple scheme (every other character becomes a space, then repeated spaces become one)
and check the result.

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
ALPHABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "               # 26 letters + space = 27 characters
char_to_id = {c: k for k, c in enumerate(ALPHABET)}   # 'A' -> 0, ..., 'Z' -> 25, ' ' -> 26

def clean_text(raw_text):
    """Upper case, replace every character not in ALPHABET by a space, then single spaces."""
    text = raw_text.upper()
    text = "".join(c if c in ALPHABET else " " for c in text)
    return " ".join(text.split())

english = clean_text(moby)         # moby = raw Moby Dick text downloaded in Core 2

# checks: only the 27 characters in ALPHABET are present, and no double spaces
assert(set(english) <= set(ALPHABET))
assert "  " not in english
print(len(english), "characters")
print(english[:300])
```

### Step 2 – Count characters and bigrams

```python
# frequency(i): number of times character i appears in the text
# (new name, so it does not clash with the letter_counts used for the ciphertext in Core 3)
char_freq = np.array([english.count(c) for c in ALPHABET])

# frequency(ij): number of times character i is immediately followed by character j
bigram_freq = np.zeros((27, 27))
for a,b in zip(english, english[1:]):       # every pair of consecutive characters
    bigram_freq[char_to_id[a], char_to_id[b]] += 1

print("pairs counted:", int(bigram_freq.sum()), " (should equal len(english) - 1 =", len(english) - 1, ")")
```

### Step 3 – Turn counts into probabilities $p(i,j)$

```python
# p(i, j) = (frequency(ij) + 1) / frequency(i)
# char_freq.reshape(-1, 1) makes the counts a column, so row i of bigram_freq is divided by frequency(i)
P = (bigram_freq + 1) / char_freq.reshape(-1, 1)

log_P = np.log(P)   # the score only needs the log probabilities, so compute it once here

def show(c):
    return "_" if c == " " else c           # print the space as "_" so it is visible to us

def most_likely_next(c, k = 3):
    # The k characters most likely to follow c, with their probabilities
    row = P[char_to_id[c]]
    best = sorted(range(27), key = lambda j: row[j], reverse=True)[:k]    # indices of the k largest values
    return ", ".join(f"{show(ALPHABET[j])} ({row[j]:.2f})" for j in best)

# Sanity check: does the model "know" some English?
for c in ["Q", "T", "H", "W", " "]:
    print(show(c), "->", most_likely_next(c))
```

```python
labels = [show(c) for c in ALPHABET]

plt.figure(figsize=(7, 6))
plt.imshow(log_P, cmap="Blues")            # one hue: darker = more likely pair
plt.xticks(range(27), labels)
plt.yticks(range(27), labels)
plt.xlabel("next character  j")
plt.ylabel("current character  i")
plt.title("log p(i, j) estimated from Moby Dick")
plt.colorbar(label="log p(i, j)")
plt.tight_layout()
plt.show()
```

### Step 4 – The score function $S(m)$
`score` follows the formula directly. `score_ids` does the same calculation on an array of integers using numpy
indexing; Core 6 evaluates $S$ hundreds of thousands of times, so this faster version is the one to use there.

```python
def score(message):
    # S(m) = sum of log p(m_j, m_{j+1}) over consecutive pairs. Larger = more natural English
    S = 0.0
    for a, b in zip(message, message[1:]):
        S += log_P[char_to_id[a], char_to_id[b]]
    return S

def to_ids(message):
    # Convert a string into an array of integers from 0 to 26
    return np.array([char_to_id[c] for c in message])

def score_ids(ids):
    # Same as score(), but for an integer array: log_P[ids[j], ids[j+1]] for all j at once
    return log_P[ids[:-1], ids[1:]].sum()

print(score("THE WHALE"), score("QXZ JVKWQ"))    # same length, very different scores

m = ciphertexts[2]
print(score(m), score_ids(to_ids(m)))             # the two versions must agree
```

### Step 5 – Comparing candidate decryptions 
For message 2 the score should increase as the decryption gets better:
random key < frequency key (Core 3) < hand-corrected key (Core 4).
We also print $S$ per pair, $S/(n-1)$, which lets us compare messages of different lengths.

```python
def decode(cipher, key):
    # key[k] is the plaintext character for cipher character ALPHABET[k] (same convention as Cores 3-4)
    return cipher.translate(str.maketrans(ALPHABET, key))

random.seed(0)                                  # reset the seed, so this cell gives the same key every time it is run
random_key = "".join(random.sample(ALPHABET, 27))
frequency_key = "HPEJSRYUKMCABFZQGIWTVL XDNO"   # Core 3: k-th most common cipher char -> k-th most common English char

c2 = ciphertexts[2]
candidates = {
    "ciphertext as it is":         c2,
    "random key":                  decode(c2, random_key),
    "frequency key (Core 3)":      decode(c2, frequency_key),
    "hand-corrected key (Core 4)": decode(c2, plain_alphabet),
}
for name, text in candidates.items():
    S = score(text)
    print(f"{name:28s} S = {S:8.0f}   S per pair = {S / (len(text) - 1):6.2f}   {text[:35]}")
```

### Unit tests for the score functions
A hand-computed example ("THE" contains exactly the bigrams TH and HE), and a check that `score` and `score_ids` give the same result on message 2.

```python
# "THE" contains exactly two bigrams: TH and HE
expected = ( log_P[char_to_id["T"], char_to_id["H"]] + log_P[char_to_id["H"], char_to_id["E"]])

assert abs(score("THE") - expected) < 1e-10

# The two implementations should give the same score
message = ciphertexts[2]
assert abs(score(message) - score_ids(to_ids(message))) < 1e-8

print("Score checks passed.")
```

### Step 6 – How Core 6 calls the score
Core 6 evaluates $S$ many times, so it should use `score_ids`. The ciphertext is converted once with `to_ids`, and the key is stored as an integer array `key_ids`, where `key_ids[k]` is the plaintext id of cipher character `k`. Then `key_ids[cipher_ids]` decodes the whole message in one step (it picks `key_ids[c]` for every cipher character `c`), and a proposal in Core 6 is a swap of two entries of `key_ids`.

```python
# Example: message 2 with the Core 4 key
cipher_ids = to_ids(ciphertexts[2])    # convert the ciphertext once
key_ids = to_ids(plain_alphabet)       # key_ids[k] = plaintext id of cipher character k
decoded_ids = key_ids[cipher_ids]      # decode the whole message in one step
S = score_ids(decoded_ids)

# Check: same text and same score as the string version in Step 5
text = decode(ciphertexts[2], plain_alphabet)
assert "".join(ALPHABET[i] for i in decoded_ids) == text
assert abs(S - score(text)) < 1e-8

print("Core 6 interface check passed. S =", round(S, 2))
```
